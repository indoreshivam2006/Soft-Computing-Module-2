### 1. Aim
To create a simple single-layer neural network using Python and train it on the AND logic gate.

### 2. Theory: Single-Layer Neural Networks

A single-layer neural network (single-layer perceptron) has only an input layer and one layer of output neurons directly connected to the inputs — **no hidden layers**.

#### 1. How It Computes
Each output neuron computes:
$$z = \mathbf{w} \cdot \mathbf{x} + b$$
$$y = \text{activation}(z)$$

#### 2. The Perceptron Learning Rule
We train this network by adjusting weights based on prediction errors:
$$e = y_{\text{target}} - y_{\text{predicted}}$$
$$w_i \leftarrow w_i + \eta \cdot e \cdot x_i$$
$$b \leftarrow b + \eta \cdot e$$

Where $\eta$ is the learning rate (how big a step we take each time).

#### 3. Limitation
A single-layer network can only solve **linearly separable** problems (like AND, OR). It **cannot** solve XOR because XOR's classes cannot be separated by a single straight line.

### 3. Python Code: Defining the Single-Layer Network Class

#### What this cell does:
In this code cell, we define a `SingleLayerNetwork` class that implements the perceptron learning rule with a step activation function.

```python
import numpy as np

class SingleLayerNetwork:
    """A single layer of perceptrons (no hidden layers)."""

    def __init__(self, num_inputs, num_outputs, learning_rate=0.1):
        self.weights = np.random.uniform(-1, 1, (num_outputs, num_inputs))
        self.bias = np.zeros(num_outputs)
        self.lr = learning_rate

    def step(self, z):
        return np.where(z >= 0, 1, 0)

    def forward(self, x):
        z = np.dot(self.weights, x) + self.bias
        return self.step(z)

    def train(self, X, y, epochs=20):
        for epoch in range(epochs):
            total_error = 0
            for xi, target in zip(X, y):
                prediction = self.forward(xi)
                error = target - prediction
                self.weights += self.lr * np.outer(error, xi)
                self.bias += self.lr * error
                total_error += np.sum(np.abs(error))
            if (epoch + 1) % 5 == 0 or epoch == 0:
                print(f"  Epoch {epoch + 1:>2}/{epochs}  -  Total Error: {total_error}")
            if total_error == 0:
                print(f"  Converged at epoch {epoch + 1}!")
                break
```

### 4. Training on the AND Gate

#### What this cell does:
In this code cell, we create training data for the AND gate (4 input combinations) and train our single-layer network on it, printing the epoch-by-epoch progress.

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [0], [0], [1]])

np.random.seed(42)
net = SingleLayerNetwork(num_inputs=2, num_outputs=1, learning_rate=0.1)

print("Training single-layer network on the AND gate:")
net.train(X, y, epochs=20)

print("\nFinal weights:", net.weights)
print("Final bias   :", net.bias)
```

### 5. Testing the Trained Network

#### What this cell does:
In this code cell, we test the trained network on all 4 input combinations to verify it has learned the AND truth table correctly.

```python
print(f"{'Input':<10} | {'Predicted':<10} | {'Expected'}")
print("-" * 36)
for xi, target in zip(X, y):
    pred = net.forward(xi)
    print(f"{str(xi):<10} | {str(pred):<10} | {target}")
```

### 6. Conclusion

The single-layer network successfully learned the AND gate because AND is a **linearly separable** problem — a single straight line in 2D can separate the output-1 region from the output-0 region. It converged in just 11 epochs with 100% accuracy.
