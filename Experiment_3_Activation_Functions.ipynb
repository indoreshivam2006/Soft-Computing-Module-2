# Experiment 3: Sigmoid, Tanh, and ReLU Activation Functions

**Objective:** To implement and visualize Sigmoid, Tanh, and ReLU activation functions using Python.

## Theory

Activation functions introduce **non-linearity** into a neural network. Without them, no matter how many layers are stacked, the network could only represent straight-line (linear) relationships — not enough for real-world problems like image or digit recognition.

1. **Sigmoid:** `f(z) = 1 / (1 + e^-z)`, Range: (0, 1) — squashes input into a probability-like value; common in output layers for binary classification.
2. **Tanh:** `f(z) = (e^z - e^-z) / (e^z + e^-z)`, Range: (-1, 1) — zero-centered, often trains faster than sigmoid in hidden layers.
3. **ReLU:** `f(z) = max(0, z)`, Range: [0, ∞) — fast to compute, avoids the vanishing-gradient problem; the most common choice for hidden layers today.

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def tanh(z):
    return np.tanh(z)

def relu(z):
    return np.maximum(0, z)
```

## Sample Values

```python
sample_inputs = np.array([-3, -1, 0, 1, 3], dtype=float)

print(f"{'z':>6} | {'Sigmoid':>10} | {'Tanh':>10} | {'ReLU':>10}")
print("-" * 46)
for z in sample_inputs:
    print(f"{z:>6.1f} | {sigmoid(z):>10.4f} | {tanh(z):>10.4f} | {relu(z):>10.4f}")
```

## Visualization

```python
z_values = np.linspace(-10, 10, 400)
y_sigmoid = sigmoid(z_values)
y_tanh = tanh(z_values)
y_relu = relu(z_values)

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

axes[0].plot(z_values, y_sigmoid, color="#3E92CC", linewidth=2.5)
axes[0].set_title("Sigmoid: f(z) = 1 / (1 + e^-z)")
axes[0].axhline(0, color="gray", linewidth=0.6); axes[0].axvline(0, color="gray", linewidth=0.6)
axes[0].set_xlabel("z"); axes[0].set_ylabel("f(z)"); axes[0].grid(alpha=0.3)

axes[1].plot(z_values, y_tanh, color="#F72585", linewidth=2.5)
axes[1].set_title("Tanh: f(z) = tanh(z)")
axes[1].axhline(0, color="gray", linewidth=0.6); axes[1].axvline(0, color="gray", linewidth=0.6)
axes[1].set_xlabel("z"); axes[1].set_ylabel("f(z)"); axes[1].grid(alpha=0.3)

axes[2].plot(z_values, y_relu, color="#161B33", linewidth=2.5)
axes[2].set_title("ReLU: f(z) = max(0, z)")
axes[2].axhline(0, color="gray", linewidth=0.6); axes[2].axvline(0, color="gray", linewidth=0.6)
axes[2].set_xlabel("z"); axes[2].set_ylabel("f(z)"); axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

## Conclusion

Sigmoid and Tanh both produce smooth S-shaped curves that saturate at the extremes (their gradients shrink to near-zero for large |z|), while ReLU stays linear for positive inputs and completely flat for negative inputs — this simplicity is exactly why ReLU trains deep networks faster in practice.
