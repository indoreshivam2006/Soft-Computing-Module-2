### 2. Theory: Why Do We Need Activation Functions?

Without activation functions, stacking layers of neurons would still only draw straight lines. Activation functions add **curves (non-linearity)** so networks can learn complex patterns like photos, voices, and medical data.

#### 1. Sigmoid Function
- **Formula:** $\sigma(z) = \frac{1}{1 + e^{-z}}$
- **Output Range:** Always between $0$ and $1$.
- **Best Use:** At the final output layer to predict a probability (like 80% chance of rain).
- **Slope Formula (Derivative):** $\sigma'(z) = \sigma(z)(1 - \sigma(z))$

#### 2. Tanh Function
- **Formula:** $\tanh(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}}$
- **Output Range:** Between $-1$ and $+1$ (centered around zero).
- **Best Use:** Hidden layers when negative numbers are helpful.
- **Slope Formula (Derivative):** $\tanh'(z) = 1 - \tanh^2(z)$

#### 3. ReLU Function (Most Popular)
- **Formula:** $\text{ReLU}(z) = \max(0, z)$
- **Output Range:** From $0$ to as high as the input goes.
- **Best Use:** Hidden layers of almost all modern deep learning models.
- **Slope Formula (Derivative):** $1$ if $z > 0$, else $0$. It is super simple and fast for computers to compute.

### 3. Python Code: Defining the Activation Functions

#### What this cell does:
In this code cell, we define all three activation functions (`sigmoid`, `tanh`, `relu`) using NumPy, along with their derivative functions for later use in backpropagation.

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def tanh(z):
    return np.tanh(z)

def relu(z):
    return np.maximum(0, z)
```

### 4. Sample Values: Computing Outputs for Selected Inputs

#### What this cell does:
In this code cell, we compute and print the output of each activation function for a set of sample input values ($z = -3, -1, 0, 1, 3$) to see how they behave at different points.

```python
sample_inputs = np.array([-3, -1, 0, 1, 3], dtype=float)

print(f"{'z':>6} | {'Sigmoid':>10} | {'Tanh':>10} | {'ReLU':>10}")
print("-" * 46)
for z in sample_inputs:
    print(f"{z:>6.1f} | {sigmoid(z):>10.4f} | {tanh(z):>10.4f} | {relu(z):>10.4f}")
```

### 5. Visualization: Plotting the Activation Functions

#### What this cell does:
In this code cell, we plot all three activation functions side by side on the same graph to visually compare their shapes, ranges, and behavior.

```python
z_values = np.linspace(-10, 10, 400)
y_sigmoid = sigmoid(z_values)
y_tanh = tanh(z_values)
y_relu = relu(z_values)

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))

axes[0].plot(z_values, y_sigmoid, color="#3E92CC", linewidth=2.5)
axes[0].set_title("Sigmoid: f(z) = 1 / (1 + e^-z)")
axes[0].axhline(0, color="gray", linewidth=0.6); axes[0].axvline(0, color="gray", linewidth=0.6)
axes[0].set_xlabel("z"); axes[0].set_ylabel("f(z)"); axes[0].grid(alpha=0.3)

axes[1].plot(z_values, y_tanh, color="#F72585", linewidth=2.5)
axes[1].set_title("Tanh: f(z) = tanh(z)")
axes[1].axhline(0, color="gray", linewidth=0.6); axes[1].axvline(0, color="gray", linewidth=0.6)
axes[1].set_xlabel("z"); axes[1].set_ylabel("f(z)"); axes[1].grid(alpha=0.3)

axes[2].plot(z_values, y_relu, color="#161B33", linewidth=2.5)
axes[2].set_title("ReLU: f(z) = max(0, z)")
axes[2].axhline(0, color="gray", linewidth=0.6); axes[2].axvline(0, color="gray", linewidth=0.6)
axes[2].set_xlabel("z"); axes[2].set_ylabel("f(z)"); axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

### 6. Conclusion

Sigmoid and Tanh both produce smooth S-shaped curves that saturate at the extremes (the gradient becomes very small), which can slow training in deep networks. ReLU is simple, fast, and avoids the vanishing-gradient problem — making it the default choice for hidden layers today.
