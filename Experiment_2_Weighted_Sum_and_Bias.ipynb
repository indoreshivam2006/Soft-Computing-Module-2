### 1. Aim
To implement the weighted sum of inputs and bias used in an artificial neuron using Python.

### 2. Theory: The Weighted Sum and Bias

Before an activation function is applied, every neuron first computes a **weighted sum** of its inputs, plus a bias term.

#### 1. The Formula
$$z = (w_1 \times x_1 + w_2 \times x_2 + \cdots + w_n \times x_n) + b = \sum_{i=1}^{n} w_i x_i + b = \mathbf{w} \cdot \mathbf{x} + b$$

#### 2. Role of Weights and Bias
- **Weights ($\mathbf{w}$):** Determine how important each input is. A large positive weight means that input strongly pushes the neuron toward firing.
- **Bias ($b$):** Shifts the result up or down, letting the neuron activate even when all inputs are zero (or stay silent even with strong inputs).

#### 3. Why Bias Matters
Without bias ($b = 0$), the decision boundary is forced to pass through the origin. The bias acts as a learnable threshold that gives the neuron more flexibility.

### 3. Python Code: Defining the Weighted Sum Functions

#### What this cell does:
In this code cell, we define two functions — `weighted_sum_manual()` that computes the weighted sum using an explicit loop (for understanding), and `weighted_sum_vectorized()` that uses NumPy's `np.dot()` for efficient computation.

```python
import numpy as np

def weighted_sum_manual(inputs, weights, bias):
    """Compute weighted sum using an explicit loop (for understanding)."""
    total = 0.0
    for x, w in zip(inputs, weights):
        total += x * w
    total += bias
    return total

def weighted_sum_vectorized(inputs, weights, bias):
    """Compute weighted sum using NumPy's dot product (efficient way)."""
    return np.dot(inputs, weights) + bias
```

### 4. Demonstration: Computing the Weighted Sum

#### What this cell does:
In this code cell, we test both methods on the same inputs, weights, and bias to verify they produce identical results.

```python
inputs = [0.6, 0.2, 0.9, 0.4]
weights = [0.3, -0.5, 0.8, 0.1]
bias = 0.25

print(f"Inputs  : {inputs}")
print(f"Weights : {weights}")
print(f"Bias    : {bias}")

z_manual = weighted_sum_manual(inputs, weights, bias)
z_vector = weighted_sum_vectorized(np.array(inputs), np.array(weights), bias)

print(f"\nWeighted sum (manual loop)       z = {z_manual:.4f}")
print(f"Weighted sum (NumPy dot product) z = {z_vector:.4f}")

assert np.isclose(z_manual, z_vector), "Mismatch between methods!"
print("\nBoth methods match, confirming: z = (w1*x1 + w2*x2 + ... + wn*xn) + b")
```

### 5. Effect of Bias

#### What this cell does:
In this code cell, we keep inputs and weights fixed and vary only the bias to see how it directly shifts the weighted sum up or down.

```python
print(f"{'Bias':>6} | {'z':>8}")
print("-" * 18)
for b in [-1.0, -0.5, 0.0, 0.5, 1.0]:
    z = weighted_sum_vectorized(np.array(inputs), np.array(weights), b)
    print(f"{b:>6.2f} | {z:>8.4f}")
```

### 6. Conclusion

The weighted sum combines all inputs into a single value `z`, and the bias shifts that value. Both the manual loop and vectorized `np.dot` approaches give identical results, but the vectorized version is much faster for large inputs.
