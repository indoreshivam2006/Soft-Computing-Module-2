### 1. Aim
To create a Multilayer Perceptron (MLP) with input, hidden, and output layers using Keras/TensorFlow to solve the XOR problem.

### 2. Theory: Why Do We Need Hidden Layers?

A single-layer perceptron (Experiment 4) can only solve **linearly separable** problems (like AND, OR) — it **cannot** solve XOR, because XOR's classes cannot be separated by a single straight line.

#### 1. The Multilayer Perceptron (MLP)
An MLP adds one or more **hidden layers** between input and output:
```
Input Layer  →  Hidden Layer(s)  →  Output Layer
(raw data)       (extract patterns)   (final prediction)
```

#### 2. Why This Works for XOR
Each hidden layer uses a non-linear activation function (e.g. ReLU), allowing the network to learn curved / complex decision boundaries — enough to solve XOR.

#### 3. Architecture Used Here
- **Input Layer:** 2 features ($x_1, x_2$)
- **Hidden Layer:** 8 Dense neurons with ReLU activation
- **Output Layer:** 1 Dense neuron with Sigmoid activation
- **Optimizer:** Adam ($\alpha = 0.05$), **Loss:** Binary Cross-Entropy

### 3. Python Code: Importing Libraries and Preparing Data

#### What this cell does:
In this code cell, we import TensorFlow/Keras and prepare the XOR dataset — 4 input pairs with their expected outputs.

````python
import os
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"

import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

np.random.seed(7)
tf.random.set_seed(7)

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
y = np.array([[0], [1], [1], [0]], dtype=float)
````

### 4. Building the MLP Architecture

#### What this cell does:
In this code cell, we build the MLP model using Keras `Sequential` API — an input layer (2 features), one hidden layer (8 neurons, ReLU), and one output layer (1 neuron, Sigmoid).

````python
model = keras.Sequential([
    layers.Input(shape=(2,)),
    layers.Dense(8, activation="relu", name="hidden_layer"),
    layers.Dense(1, activation="sigmoid", name="output_layer"),
])

optimizer = keras.optimizers.Adam(learning_rate=0.05)
model.compile(optimizer=optimizer, loss="binary_crossentropy", metrics=["accuracy"])

model.summary()
````

### 5. Training on the XOR Problem

#### What this cell does:
In this code cell, we compile and train the model for 500 epochs on XOR data, then evaluate the final accuracy.

````python
model.fit(X, y, epochs=500, verbose=0)

loss, accuracy = model.evaluate(X, y, verbose=0)
print(f"Final training accuracy: {accuracy * 100:.2f}%")
print(f"Final training loss    : {loss:.4f}")
````

### 6. Predictions: Testing on All Inputs

#### What this cell does:
In this code cell, we pass all 4 XOR inputs through the trained model and print the predicted probabilities alongside the expected values.

````python
print(f"{'Input':<10} | {'Predicted':<12} | {'Rounded':<8} | {'Expected'}")
print("-" * 50)
predictions = model.predict(X, verbose=0)
for xi, pred, target in zip(X, predictions, y):
    print(f"{str(xi.astype(int)):<10} | {pred[0]:<12.4f} | {round(pred[0]):<8} | {int(target[0])}")
````

### 7. Conclusion

The MLP's hidden layer allows it to solve XOR with 100% accuracy — something a single-layer perceptron cannot do. This demonstrates the power of **hidden layers** and **non-linear activations** in learning complex, non-linearly separable patterns.
