## Practical 3: Activation Functions (Sigmoid, Tanh, ReLU)

### Aim
To implement and visualize Sigmoid, Tanh, and ReLU activation functions using Python.

### Theory
Activation functions introduce non-linearity into neural networks, enabling them to learn complex non-linear patterns. The three fundamental activation functions are:

1. **Sigmoid:**  
   $$\sigma(x) = \frac{1}{1 + e^{-x}}, \quad \text{Range: } (0, 1)$$
2. **Tanh (Hyperbolic Tangent):**  
   $$\tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}, \quad \text{Range: } (-1, 1)$$
3. **ReLU (Rectified Linear Unit):**  
   $$\text{ReLU}(x) = \max(0, x), \quad \text{Range: } [0, \infty)$$

### Implementation
Defining and computing Sigmoid, Tanh, and ReLU functions:

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

# Test with sample values
sample_inputs = np.array([-3.0, -1.0, 0.0, 1.0, 3.0])
print("Inputs :", sample_inputs)
print("Sigmoid:", np.round(sigmoid(sample_inputs), 3))
print("Tanh   :", np.round(tanh(sample_inputs), 3))
print("ReLU   :", np.round(relu(sample_inputs), 3))
```

### Visualization
Plotting and comparing all three activation functions over a continuous range:

```python
x = np.linspace(-5, 5, 200)

plt.figure(figsize=(9, 3.5))

plt.subplot(1, 3, 1)
plt.plot(x, sigmoid(x), color='blue', linewidth=2)
plt.title('Sigmoid: (0, 1)')
plt.grid(True, alpha=0.3)

plt.subplot(1, 3, 2)
plt.plot(x, tanh(x), color='green', linewidth=2)
plt.title('Tanh: (-1, 1)')
plt.grid(True, alpha=0.3)

plt.subplot(1, 3, 3)
plt.plot(x, relu(x), color='red', linewidth=2)
plt.title('ReLU: [0, inf)')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Observation
- **Sigmoid** squashes values between 0 and 1, useful for binary output probabilities.
- **Tanh** is zero-centered with outputs between -1 and 1, making gradient descent converge faster in hidden layers.
- **ReLU** avoids vanishing gradients for positive inputs and is computationally very fast.
