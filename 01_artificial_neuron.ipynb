## Practical 1: Basic Working of an Artificial Neuron

### Aim
To understand and implement the basic working of an artificial neuron using Python.

### Theory
An **Artificial Neuron** (inspired by biological neurons) takes several input signals $x_1, x_2, \dots, x_n$, multiplies each by an associated weight $w_i$, adds a bias term $b$, and computes the **net input** $z$:

$$z = \sum_{i=1}^{n} w_i x_i + b = \mathbf{w}^T \mathbf{x} + b$$

The net input is then passed through an **activation function** $f(z)$ (such as Sigmoid) to generate the final neuron output $y$:

$$y = f(z) = \frac{1}{1 + e^{-z}}$$

### Implementation
A concise implementation of a single artificial neuron using NumPy:

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define inputs, weights, and bias
X = np.array([2.5, 1.5])    # Two input features
W = np.array([0.8, -0.5])   # Respective weights
b = 0.2                     # Bias term

# 2. Calculate net input: z = w1*x1 + w2*x2 + b
z = np.dot(X, W) + b

# 3. Activation function (Sigmoid)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

output = sigmoid(z)

print(f"Inputs (X)        : {X}")
print(f"Weights (W)       : {W}")
print(f"Bias (b)          : {b}")
print(f"Net Input (z)     : {z:.4f}")
print(f"Neuron Output (y) : {output:.4f}")
```

### Visualization
Plotting the sigmoid activation curve and showing our neuron's operating point:

```python
# Plotting the sigmoid curve and our neuron output point
z_vals = np.linspace(-6, 6, 100)
plt.figure(figsize=(6, 3.5))
plt.plot(z_vals, sigmoid(z_vals), label='Sigmoid Function', color='blue')
plt.scatter([z], [output], color='red', s=80, zorder=5, label=f'Neuron Output (z={z:.2f}, y={output:.2f})')
plt.title('Artificial Neuron Activation')
plt.xlabel('Net Input (z)')
plt.ylabel('Activation Output f(z)')
plt.axhline(0.5, color='gray', linestyle='--', alpha=0.5)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Observation
- For inputs $[2.5, 1.5]$ with weights $[0.8, -0.5]$ and bias $0.2$, the net input $z = (2.5 \times 0.8) + (1.5 \times -0.5) + 0.2 = 1.45$.
- After applying the sigmoid function, the output is approximately $0.8099$, which is greater than $0.5$ (indicating an active firing state).
