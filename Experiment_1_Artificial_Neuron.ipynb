# Experiment 1: Basic Working of an Artificial Neuron

**Objective:** To understand and implement the basic working of an artificial neuron using Python.

## Theory

An artificial neuron (perceptron) is the fundamental building block of a Neural Network, loosely inspired by a biological neuron:

| Biological Neuron | Artificial Neuron |
|---|---|
| Dendrites (receive signal) | Inputs (x1, x2, ..., xn) |
| Synapse strength | Weights (w1, w2, ..., wn) |
| Cell body (sums signals) | Weighted sum + bias |
| Firing threshold | Activation function |
| Axon (sends output) | Output (y) |

A single neuron performs two steps:
1. Compute the **weighted sum** of inputs plus a bias: `z = (w . x) + b`
2. Pass `z` through an **activation function** to produce the output.

Here we use a **step activation function** — the neuron "fires" (outputs 1) if `z >= 0`, else stays silent (outputs 0).

```python
import numpy as np

class ArtificialNeuron:
    """A single artificial neuron with a step activation function."""

    def __init__(self, num_inputs, weights=None, bias=0.0):
        if weights is None:
            self.weights = np.random.uniform(-1, 1, num_inputs)
        else:
            self.weights = np.array(weights, dtype=float)
        self.bias = bias

    def step_activation(self, z):
        """Step function: fires (1) if z >= 0, else stays silent (0)."""
        return 1 if z >= 0 else 0

    def forward(self, inputs):
        """Compute the neuron's output for a given input vector."""
        inputs = np.array(inputs, dtype=float)
        z = np.dot(self.weights, inputs) + self.bias
        output = self.step_activation(z)
        return z, output
```

## Demonstration

Create a neuron with 3 inputs and fixed weights/bias for clarity, then test it on a few input vectors.

```python
neuron = ArtificialNeuron(num_inputs=3, weights=[0.5, -0.6, 0.8], bias=0.1)

print(f"Weights : {neuron.weights}")
print(f"Bias    : {neuron.bias}")

test_inputs = [
    [1, 0, 1],
    [0, 1, 0],
    [1, 1, 1],
    [-1, -1, -1],
]

print("\nInput Vector      -> Weighted Sum (z)  -> Output")
print("-" * 55)
for inputs in test_inputs:
    z, output = neuron.forward(inputs)
    print(f"{str(inputs):<18}-> {z:>8.4f}            -> {output}")
```

## Conclusion

The neuron computes `z = (weights . inputs) + bias`, and fires (outputs 1) only when `z` crosses the activation threshold (0). This simple mechanism — weighted combination + thresholding — is the foundation every larger neural network is built from.
