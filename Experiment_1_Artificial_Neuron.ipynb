### 2. Theory: How an Artificial Neuron Works

An artificial neuron is a simple math model inspired by how our brain works.

#### 1. Brain Cell vs Computer Neuron
- **In our brain:** A brain cell receives signals from other cells, adds them together, and if the total signal is strong enough, passes it forward to the next cell.
- **In computers:** An artificial neuron takes input numbers ($x_1, x_2, \ldots, x_n$), multiplies each by a weight ($w_1, w_2, \ldots, w_n$) to decide how important it is, adds a bias number ($b$) to shift the result, and checks if the final total is big enough to output `1` (ON) or `0` (OFF).

| Biological Neuron | Artificial Neuron |
|---|---|
| Dendrites (receive signal) | Inputs ($x_1, x_2, \ldots, x_n$) |
| Synapse strength | Weights ($w_1, w_2, \ldots, w_n$) |
| Cell body (sums signals) | Weighted sum + bias |
| Firing threshold | Activation function |
| Axon (sends output) | Output ($y$) |

#### 2. The Weighted Sum Formula
$$\text{Total Sum } (z) = (w_1 \times x_1) + (w_2 \times x_2) + \cdots + (w_n \times x_n) + b$$
- $x_1, x_2, \ldots, x_n$: The input numbers.
- $w_1, w_2, \ldots, w_n$: The weights — they decide how much each input affects the total.
- $b$: The bias number — shifts the total sum up or down so the neuron can turn ON or OFF at the right time.

#### 3. The Step Rule (Activation Function)
$$f(z) = \begin{cases} 1 & \text{if } z \ge 0 \\ 0 & \text{if } z < 0 \end{cases}$$
- If Total Sum ($z$) is 0 or greater, output is **1** (Neuron fires / turns ON).
- If Total Sum ($z$) is less than 0, output is **0** (Neuron stays OFF).

### 3. Python Code: Defining the Artificial Neuron Class

#### What this cell does:
In this code cell, we define our `ArtificialNeuron` class. It takes weights and a bias, computes the weighted sum $z = \mathbf{w}^T \mathbf{x} + b$, and applies the step activation function.

```python
import numpy as np

class ArtificialNeuron:
    """A single artificial neuron with a step activation function."""

    def __init__(self, num_inputs, weights=None, bias=0.0):
        if weights is None:
            self.weights = np.random.uniform(-1, 1, num_inputs)
        else:
            self.weights = np.array(weights, dtype=float)
        self.bias = bias

    def step_activation(self, z):
        """Step function: fires (1) if z >= 0, else stays silent (0)."""
        return 1 if z >= 0 else 0

    def forward(self, inputs):
        """Compute the neuron's output for a given input vector."""
        inputs = np.array(inputs, dtype=float)
        z = np.dot(self.weights, inputs) + self.bias
        output = self.step_activation(z)
        return z, output
```

### 4. Demonstration: Testing the Neuron

#### What this cell does:
In this code cell, we create a neuron with 3 inputs and fixed weights/bias, then test it on several sample inputs to see when it fires (outputs 1) and when it stays silent (outputs 0).

```python
neuron = ArtificialNeuron(num_inputs=3, weights=[0.5, -0.6, 0.8], bias=0.1)

print(f"Weights : {neuron.weights}")
print(f"Bias    : {neuron.bias}")

test_inputs = [
    [1, 0, 1],
    [0, 1, 0],
    [1, 1, 1],
    [-1, -1, -1],
]

print("\nInput Vector      -> Weighted Sum (z)  -> Output")
print("-" * 55)
for inputs in test_inputs:
    z, output = neuron.forward(inputs)
    print(f"{str(inputs):<18}-> {z:>8.4f}            -> {output}")
```

### 5. Conclusion

The neuron computes `z = (weights · inputs) + bias`, and fires (outputs 1) only when `z ≥ 0`.
By choosing different weights and bias values, a single neuron can implement simple decision boundaries — the foundation of all neural networks.
