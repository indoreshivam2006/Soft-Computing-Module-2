# Experiment 6: Gradient Descent for Weight Updates

**Objective:** To apply Gradient Descent for updating the weights of a simple neural network during training.

## Theory

Gradient Descent is the optimization algorithm neural networks use to learn:

1. Start with random weights.
2. Measure how wrong the prediction is (the **loss**).
3. Compute the **gradient** of the loss with respect to each weight.
4. Update each weight in the **opposite** direction of its gradient, scaled by a **learning rate**:
   ```
   w_new = w_old - (learning_rate * gradient)
   ```
5. Repeat until the loss is minimized.

Here we train a single linear neuron (`y = w*x + b`) to fit a small dataset.

````python
import numpy as np

def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

# Dataset: y = 3x + 2 (plus a little noise)
np.random.seed(1)
X = np.array([1, 2, 3, 4, 5], dtype=float)
y_true = 3 * X + 2 + np.random.normal(0, 0.2, size=X.shape)

w = np.random.randn()
b = np.random.randn()
learning_rate = 0.01
epochs = 200
n = len(X)

print(f"Initial weight (w): {w:.4f}")
print(f"Initial bias   (b): {b:.4f}")
````

## Gradient Descent Loop

````python
print(f"{'Epoch':>6} | {'Loss (MSE)':>12} | {'w':>8} | {'b':>8}")
print("-" * 42)

for epoch in range(epochs):
    y_pred = w * X + b
    loss = mean_squared_error(y_true, y_pred)

    dw = (-2 / n) * np.sum(X * (y_true - y_pred))
    db = (-2 / n) * np.sum(y_true - y_pred)

    w -= learning_rate * dw
    b -= learning_rate * db

    if epoch % 20 == 0 or epoch == epochs - 1:
        print(f"{epoch:>6} | {loss:>12.4f} | {w:>8.4f} | {b:>8.4f}")

print(f"\nFinal learned weight (w): {w:.4f}   (true value ~ 3.0)")
print(f"Final learned bias   (b): {b:.4f}   (true value ~ 2.0)")
````

## Conclusion

Gradient descent progressively reduced the loss by nudging `w` and `b` toward the values that best fit the data — the learned weight and bias converge close to the true values (3 and 2) used to generate the dataset.
