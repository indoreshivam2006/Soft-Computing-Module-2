## Practical 4: Single-Layer Neural Network (Perceptron)

### Aim
To create a simple single-layer neural network using Python.

### Theory
A **Single-Layer Perceptron** consists of input nodes directly connected to an output node with a step activation function. It updates weights using the perceptron learning rule:

$$w_i \leftarrow w_i + \alpha \cdot (y - \hat{y}) \cdot x_i$$
$$b \leftarrow b + \alpha \cdot (y - \hat{y})$$

where $\alpha$ is the learning rate, $y$ is the actual label, and $\hat{y}$ is the predicted class (0 or 1).

### Implementation
Training a single-layer perceptron to learn the logical AND gate:

```python
import numpy as np
import matplotlib.pyplot as plt

# AND Gate truth table
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])

# Initialize parameters
weights = np.zeros(2)
bias = 0.0
lr = 0.1
epochs = 10

# Perceptron training loop
for epoch in range(epochs):
    for i in range(len(X)):
        net_input = np.dot(X[i], weights) + bias
        pred = 1 if net_input >= 0 else 0
        error = y[i] - pred
        weights += lr * error * X[i]
        bias += lr * error

print(f"Trained Weights : {weights}")
print(f"Trained Bias    : {bias:.2f}")

# Test predictions
predictions = [1 if np.dot(x, weights) + bias >= 0 else 0 for x in X]
print(f"Target Outputs  : {y.tolist()}")
print(f"Model Predictions: {predictions}")
```

### Visualization
Plotting the data points and the learned linear decision boundary:

```python
plt.figure(figsize=(5.5, 4))
for i in range(len(X)):
    color = 'green' if y[i] == 1 else 'red'
    plt.scatter(X[i, 0], X[i, 1], color=color, s=100, zorder=5)

# Decision boundary: w0*x1 + w1*x2 + b = 0 => x2 = -(w0*x1 + b)/w1
x1_vals = np.linspace(-0.2, 1.2, 50)
x2_vals = -(weights[0] * x1_vals + bias) / weights[1]
plt.plot(x1_vals, x2_vals, 'b--', label='Learned Decision Boundary')

plt.title('Single-Layer Perceptron: AND Gate')
plt.xlabel('Input x1')
plt.ylabel('Input x2')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Observation
- The single-layer perceptron successfully learned the weights and bias within 10 epochs.
- The decision boundary cleanly separates the $(1, 1)$ positive class from the other three points.

### Conclusion
Built and trained a simple single-layer perceptron network in Python, demonstrating linear separability and decision boundary formation.
