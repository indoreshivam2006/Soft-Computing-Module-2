## Practical 2: Weighted Sum of Inputs and Bias

### Aim
To implement the weighted sum of inputs and bias used in an artificial neuron using Python.

### Theory
The **weighted sum** represents the total incoming excitation of a neuron. Each input feature $x_i$ is scaled by its synaptic weight $w_i$, reflecting the feature's importance:

$$z = \sum_{i=1}^{n} w_i x_i + b$$

The **bias** $b$ acts as an offset or threshold. It shifts the linear combination left or right, allowing the neuron to activate even when all input values are zero.

### Implementation
Calculating the weighted sum and bias across multiple input samples using matrix dot product:

```python
import numpy as np
import matplotlib.pyplot as plt

# Define a batch of 4 samples (3 features each)
X = np.array([
    [1.0, 2.0, 3.0],
    [2.0, 0.5, 1.5],
    [0.0, 1.0, 2.0],
    [3.0, 3.0, 1.0]
])

weights = np.array([0.5, -0.2, 0.8])
bias = 1.2

# Compute net input for each sample: z = X @ weights + bias
weighted_sum = np.dot(X, weights) + bias

print("Weights:", weights)
print("Bias   :", bias)
print("\nResults for each sample:")
for i, (sample, ws) in enumerate(zip(X, weighted_sum), 1):
    print(f"  Sample {i}: X = {sample} -> Net Input (z) = {ws:.3f}")
```

### Visualization: Effect of Bias
Visualizing how changing the bias value shifts the activation boundary along the input axis:

```python
# Visualizing how different bias values shift the line z = w*x + b
x_vals = np.linspace(-4, 4, 100)
w = 1.5
biases = [-2.0, 0.0, 2.0]

plt.figure(figsize=(6.5, 3.5))
for b_val in biases:
    plt.plot(x_vals, w * x_vals + b_val, label=f'Bias b = {b_val}')

plt.title('Effect of Bias on Net Input (z = w*x + b)')
plt.xlabel('Input (x)')
plt.ylabel('Net Input (z)')
plt.axhline(0, color='black', linestyle='--', linewidth=0.8)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Observation
- The vector dot product `np.dot(X, weights)` efficiently computes the sum of products for all samples.
- Adding the bias shifts the net input vertically: positive bias increases activation, while negative bias suppresses it.
